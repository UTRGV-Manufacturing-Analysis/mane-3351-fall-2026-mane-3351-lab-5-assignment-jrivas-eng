```python
# Cell 2 (code)
import math
# Enter import code below this comment line
def logistic_cdf(x, mu, sigma):
    F = math.exp((x - mu) / sigma) / (1 + math.exp((x - mu) / sigma))
    return F
    
# Enter your code to create the user-defined function below this comment line
x = float(input("Enter the value of x: "))
mu = float(input("Enter the value of mu: "))
sigma = float(input("Enter the value of sigma: "))

# Enter your code to input values for x, mu and sigma below this comment line
F = logistic_cdf(x, mu, sigma)


# Enter the remaining code to call your function and produce the output below this comment line\
print("The value of F(x) is:", F)

```
